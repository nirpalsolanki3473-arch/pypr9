```python
print("======================== Data Analysis & Visualization program ========================")
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
class SalesDataAnalyzer:
    def __init__(self):
        self.od=None
        self.df=None

    def load_data(self):

        self.od=input('enter the path of the dataset (CSV file):')
        try:
            self.df=pd.read_csv(self.od)
            print('== Load Dataset ==')
            print('Dataset loaded successfully!')

        except Exception as err:
            print(f'sorry there is error {err}')


    def explore_data(self):
        print('== Explore Data ==')
        print('1.Display the first 5 rows \n2.Display the last 5 rows \n3.Display column names')
        print('4.Display data types \n5.Display basic info')

        if self.df is not None:
            ed=input('enter your choice')
            if ed=='1':
                print(self.df.head())

            elif ed =='2':
                print(self.df.tail())

            elif ed =='3':
                print(self.df.columns)

            elif ed =='4':
                print(self.df.dtypes)

            elif ed =='5':
                self.df.info()

        else:
            print('no dataset found')

    def operation(self):
        if self.df is not None:
            print(self.df.columns)
            print('1.*\n2.+\n3.-\n4./')
            pi=input('enter which column :')
            ip=input('Choose your choice :')
            pi2=input('enter which column :')

            if ip=='*':
                a=self.df[pi] * self.df[pi2]
                print(a)
            elif ip=='+':
                a1=self.df[pi] + self.df[pi2]
                print(a1)
            elif ip=='-':
                a2=self.df[pi] - self.df[pi2]
                print(a2)
            elif ip=='/':
                a3=self.df[pi] / self.df[pi2]
                print(a3)
            else:
                print('invalid')

        else:
            print('sorry ,file not found')

    def missing(self):
        if self.df is not None:
            print('1.Display rows with missing values \n2.Fill missing values with mean \n3.Drop rows with mising values \n4.Replace missing values with a specific value')
            ip1=input('enter your choice :')
            if self.df.isnull().sum().sum()>0 :
                if ip1=='1':
                    print(self.df[self.df.isnull().any(axis=1)])
        
                elif ip1=='2':
                    o=self.df.mean()
                    self.df=self.df.fillna(o)

                elif ip1=='3':
                    self.df.dropna(inplace=True)

                elif ip1=='4':
                    pi=input('enter value')
                    self.df=self.df.fillna(pi)
                else:
                    print('invalid choice')
            elif self.df.isnull().sum().sum()<=0:
                print('no missing value found')
        else:
            print('file not found')

    def Statistics(self):
        if self.df is not None:
            print('1.describe \n2.standard deviation \n3.variance \n4.Percentile / Quantile')
            h=input("enter your choice :")
            if h=='1':
                b=self.df.describe()
                print(b)

            elif h=='2':
                b1=self.df.std()
                print(b1)

            elif h=='3':
                b2=self.df.var()
                print(b2)

            elif h=='4':
                b3=self.df.quantile(0.50)
                print(b3)

            else:
                print('invalid choice')
        else:
            print('file not found')

    def Visualization(self):
        if self.df is not None:
            print('1.Bar Plot \n2.Line Plot \n3.Scatter Plot \n4.Pie Chart \n5.Histrogran \n6.Stack Plot \n7.boxplot \n8.heatmap')
            k=input('enter your choice')

            if k=='1':
                x=input('enter x axis column name:')
                y=input('enter y axis column name:')

                plt.bar(self.df[x],self.df[y])
                plt.show()
            elif k=='2':
                x=input('enter x axis column name:')
                y=input('enter y axis column name:')

                plt.plot(self.df[x],self.df[y])
                plt.show()
            elif k=='3':
                x=input('enter x axis column name:')
                y=input('enter y axis column name:')

                plt.scatter(self.df[x],self.df[y])
                plt.show()
            elif k=='4':
                x=input('enter column name:')
                l=input('enter labels :')

                plt.pie(self.df[x],labels=self.df[l],autopct='%1.1f%%')
                plt.show()
            elif k=='5':
                x=input('enter x axis column name:')
                y=int(input('enter number of bins:'))

                plt.hist(self.df[x],bins=y)
                plt.show()

            elif k=='6':
                x=input('enter x axis column name:')
                y=input('enter y axis column names saperated by space:').split()
                plt.stackplot(self.df[x],self.df[y])
                plt.show()

            elif k=='7':
                x=input('enter numeric column name:')
                sns.boxplot(self.df[x])
                plt.show()

            elif k=='8':
                corr=self.df.select_dtypes(include='number').corr()
                sns.heatmap(corr,annot=True,cmap='coolwarm')
                plt.show()


            else:
                print('invalid')

        else:
            print('file not found')

    def save(self):
        q=input('enter file name to save the plot (e.g., scatter_plot.png):')
        plt.savefig(q)

    def sd(self):
        cl=input('enter column name :')
        t=self.df=self.df.sort_values(cl,ascending=True)
        print(t)

    def sch(self):
        cl=input('enter column name :')
        se=input('enter what you want to search')
        print(self.df[self.df[cl] == se])

    def flt(self):
        print('1.greater than \n2.less than \n3.equal')
        mk=input('enter your choice')
        if mk=='1':
            cl=input('enter column name')
            vl=int(input('enter value'))
            l2=self.df[self.df[cl]>vl]
            print(l2)
        elif mk=='2':
            cl=input('enter column name')
            vl=int(input('enter value'))
            l1=self.df[self.df[cl]<vl]
            print(l1)

        elif mk=='3':
            cl=input('enter column name')
            vl=int(input('enter value'))
            l=self.df[self.df[cl]==vl]
            print(l)
        else:
            print('invalid')

    def agg(self):
        cl=input('enter numeric column name')
        ik=input('enter what you want 1.sum or 2.mean or 3.median')
        if ik=='1':
            s=self.df[cl].sum()
            print(s)
        
        elif ik=='2':
            s=self.df[cl].mean()
            print(s)
        
        elif ik=='3':
            s=self.df[cl].median()
            print(s)
        else:
            print('invalid')

    def gb(self):
        x1=input('enter category column')
        x2=input('enter numeric column')
        xx=input('what you want 1.sum or 2.mean or 3.median')
        if xx=='1':
            s1=self.df.groupby(x1)[x2].sum()
            print(s1)
        elif xx=='2':
            s2=self.df.groupby(x1)[x2].mean()
            print(s2)
        elif xx=='3':
            s3=self.df.groupby(x1)[x2].median()
            print(s3)
        else:
            print('invalid')
    
    def ptable(self):
        ai=input('enter index')
        ac=input('enter column')
        av=input('enter values')

        x=pd.pivot_table(index=self.df[ai],columns=self.df[ac],values=self.df[av])
        print(x)

    def splot(self):
        an=input('enter 1st numeric column')
        ad=input('enter 2nd numeric column')

        fig,axs=plt.subplots(2,2,figsize=(10,6))

        axs[0][0].plot(self.df[an],self.df[ad])
        axs[0][1].barh(self.df[an],self.df[ad])
        axs[1][0].bar(self.df[an],self.df[ad])
        axs[1][1].scatter(self.df[an],self.df[ad])

        plt.show()

    def add(self):
        print('add Data Frame , 1.below datframe 2.index wise 3.column wise')
        cc=input('enter your choice')
        cc1=input('enter dataframe name:')
        if cc=='1':
            x=pd.concat([self.df,pd.read_csv(cc1)])
            print(x)

        elif cc=='2':
            dff=self.df.join(pd.read_csv(cc1))
            print(dff)

        elif cc=='3':
            er=input('enter column name :')
            dt=pd.merge(self.df,pd.read_csv(cc1),on=er)
            print(dt)

        else:
            print('invalid')

    def nump(self):
        if self.df is not None:
            print(self.df.columns)
            cg=input('enter numeric column name :')
            if cg in self.df.columns:
                ni=np.array(self.df[cg])

                ee=input('enter your choice , 1.maximum value 2.minimum value 3.average value 4.indexing 5.slicing')
                if ee=='1':
                    b=ni.max()
                    print(b)
                elif ee=='2':
                    b1=ni.min()
                    print(b1)
                elif ee=='3':
                    b2=ni.mean()
                    print(b2)
                elif ee=='4':
                    eei=int(input('enter index'))
                    v=ni[eei]
                    print(v)
                elif ee=='5':
                    ee1=int(input('enter start index :'))
                    ee4=int(input('enter end index :'))

                    v1=ni[ee1:ee4]
                    print(v1)
                else:
                    print('invalid')

            else:
                print('not found')

        else:
            print('not found')


ld=SalesDataAnalyzer()

while True:
    print('please select an option :')
    print('1.Load Dataset \n2.Explore Data \n3.Perform DataFrame Operations \n4.Handle Missing Data')
    print('5.Generate Descriptive Statistics \n6.Data Visualization \n7.sort \n8.search \n9.filter \n10.aggregate \n11.group by \n12.Save Visualization \n13.pivot table \n14.subplots \n15.add dataframe \n16.numpy \n17.Exit')

    o=input('enter your choice :')
    if o=='1':
        ld.load_data()

    elif o=='2':
        ld.explore_data()

    elif o=='3':
        ld.operation()

    elif o=='4':
        ld.missing()

    elif o=='5':
        ld.Statistics()

    elif o=='6':
        ld.Visualization()

    elif o=='7':
        ld.sd()

    elif o=='8':
        ld.sch()

    elif o=='9':
        ld.flt()
        
    elif o=='10':
        ld.agg()

    elif o=='11':
        ld.gb()

    elif o=='12':
        ld.save()

    elif o=='13':
        ld.ptable()

    elif o=='14':
        ld.splot()

    elif o=='15':
        ld.add()

    elif o=='16':
        ld.nump()

    elif o=='17':
        break





    else:
        print('invalid')
```
